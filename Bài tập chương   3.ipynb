```python
#3.2
diem_goc = [7.5, 9.0, 3.0, 8.0, 6.5, 10.0, 7.0, 2.5]

ds = []
for val in diem_goc:
    ds.append(val)

n_elem = len(ds)
for i in range(n_elem):
    for j in range(0, n_elem - i - 1):
        if ds[j] > ds[j + 1]:
        
            ds[j], ds[j + 1] = ds[j + 1], ds[j]

n = 2
ds_cat = []
for i in range(n, len(ds) - n):
    ds_cat.append(ds[i])

tong_cat = 0
for val in ds_cat:
    tong_cat += val
tb_cat = tong_cat / len(ds_cat) if ds_cat else 0

tong_goc = 0
for val in diem_goc:
    tong_goc += val
tb_thuong = tong_goc / len(diem_goc)
print(f"Danh sách sau khi sắp xếp nổi bọt và cắt: {ds_cat}")
print(f"Trung bình cắt đầu - cuối: {tb_cat:.2f}")
```

```python
#3.4
du_lieu = [(1, 4.0), (2, 5.5), (3, 6.0), (4, 7.5), (5, 8.0)]

x_list = []
y_list = []
for xi, yi in du_lieu:
    x_list.append(xi)
    y_list.append(yi)

n = len(du_lieu)
x_bar = sum(x_list) / n
y_bar = sum(y_list) / n
tu_so = 0
mau_so = 0
for xi, yi in zip(x_list, y_list):
    tu_so += (xi - x_bar) * (yi - y_bar)
    mau_so += (xi - x_bar) ** 2

b = tu_so / mau_so
a = y_bar - b * x_bar

gio_on = 6
diem_du_doan = a + b * gio_on

print("\n  Hồi quy tuyến tính")
print(f"Phương trình hồi quy: y = {a:.4f} + {b:.4f} * x")
print(f"Dự đoán điểm cho sinh viên ôn {gio_on} giờ: {diem_du_doan:.2f}")
```

```python
#3.6
danh_sach_sv = [
    ("An", 9.0), ("Bình", 7.5), ("Chi", 8.6), 
    ("Dung", 5.0), ("Em", 6.0), ("Phúc", 3.5), ("Giang", 7.0)
]

def quy_doi(diem):
    if diem >= 8.5: return 'A'
    elif diem >= 7.0: return 'B'
    elif diem >= 5.5: return 'C'
    elif diem >= 4.0: return 'D'
    else: return 'F'

nhom_sv = {}
for ten, diem in danh_sach_sv:
    diem_chu = quy_doi(diem)
    nhom_sv.setdefault(diem_chu, []).append(ten)

print("\n  Gom nhóm học lực sinh viên")
for key in sorted(nhom_sv.keys()):
    print(f"Điểm {key}: {nhom_sv[key]}")
```

```python
#3.8
cua_hang_1 = {"ao": 120, "quan": 80, "non": 30}
cua_hang_2 = {"ao": 90, "giay": 150, "non": 20}
tong_doanh_thu = {}
for k, v in cua_hang_1.items():
    tong_doanh_thu[k] = v
for k, v in cua_hang_2.items():

    tong_doanh_thu[k] = tong_doanh_thu.get(k, 0) + v

ds_giam_dan = sorted(tong_doanh_thu.items(), key=lambda item: item[1], reverse=True)

set_1 = set(cua_hang_1.keys())
set_2 = set(cua_hang_2.keys())

ban_ca_hai = set_1 & set_2                
chi_ban_mot_noi = set_1 ^ set_2        

print("\n Phân tích doanh thu cửa hàng ")
print("Doanh thu giảm dần:", ds_giam_dan)
print("Sản phẩm bán ở cả hai cửa hàng:", ban_ca_hai)
print("Sản phẩm chỉ bán ở một cửa hàng:", chi_ban_mot_noi)
```

```python
#3.10
N = 1000
ung_vien = set(range(2, N + 1))

for i in range(2, int(N**0.5) + 1):
    if i in ung_vien:
      
        boi_so = set(range(i * 2, N + 1, i))
     
        ung_vien -= boi_so

so_nguyen_to = ung_vien
so_luong = len(so_nguyen_to)
tong_gia_tri = sum(so_nguyen_to)


cap_sinh_doi = []
for p in sorted(so_nguyen_to):
    if (p + 2) in so_nguyen_to:
        cap_sinh_doi.append((p, p + 2))

print("\nSàng Eratosthenes - tìm nguyên tố  ")
print(f"Số lượng số nguyên tố <= 1000: {so_luong}")
print(f"Tổng giá trị: {tong_gia_tri}")
print(f"Số lượng cặp sinh đôi: {len(cap_sinh_doi)}")
```
